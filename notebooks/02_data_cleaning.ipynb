### Correct Misplaced Duration Values

**Problem:** Three movie durations (74 min, 84 min, 66 min) were incorrectly stored in the `rating` column.

**Cleaning strategy:**
- Move duration values from `rating` to `duration`.
- Replace incorrect rating values with `None`.

**Code:**

```python
mask = df['rating'].str.contains('min', na=False)

df.loc[mask, 'duration'] = df.loc[mask, 'rating']
df.loc[mask, 'rating'] = None
```

**Reason:** Maintain data consistency by ensuring movie durations are stored in the correct column.

````python
mask = df['rating'].str.contains(
    r'\bmin\b|\bSeasons?\b',
    case=False,
    na=False
)

df.loc[mask, 'duration'] = df.loc[mask, 'rating']
df.loc[mask, 'rating'] = None
df[df['rating'].isnull()][['title', 'rating', 'duration']]
````
